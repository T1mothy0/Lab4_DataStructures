```python
from collections import defaultdict

LAST_NAME = "MEJOS"
SEED_NUM = 5
FAVORITE_ARTIST = "TAYLOR SWIFT"
SUBJECTS = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

# Generate records as a list of tuples: (student, subject, score)
records = []
for i in range(5):
    student = f"{LAST_NAME}_S{i + 1}"
    score = (ord(FAVORITE_ARTIST[i]) * SEED_NUM) % 101
    records.append((student, SUBJECTS[i], score))

records.append((f"{LAST_NAME}_S1", "PROGRAMMING", 16))   # repeated record
records.append((f"{LAST_NAME}_S2", "MATHEMATICS", None))  # incomplete record

print("=== GENERATED RECORDS ===")
for r in records:
    print(r)

# Organize scores by student using a dictionary of lists
by_student = defaultdict(list)
for student, subject, score in records:
    by_student[student].append(score)

# Find repeated and incomplete records using a set for seen keys
seen = set()
repeated = []
incomplete = []
for student, subject, score in records:
    key = (student, subject)
    if key in seen:
        repeated.append((student, subject, score))
    else:
        seen.add(key)
    if score is None:
        incomplete.append((student, subject, score))

print("\n=== REPEATED ===", repeated)
print("=== INCOMPLETE ===", incomplete)

# Classify each valid score
print("\n=== CATEGORIES ===")
for student, subject, score in records:
    if score is None:
        category = "INCOMPLETE"
    elif score < 60:
        category = "NEEDS IMPROVEMENT"
    elif score < 80:
        category = "SATISFACTORY"
    else:
        category = "EXCELLENT"
    print(f"{student} - {subject}: {score} -> {category}")

# Student averages (skip missing scores)
student_avg = {}
for student, scores in by_student.items():
    valid = [s for s in scores if s is not None]
    student_avg[student] = round(sum(valid) / len(valid), 2) if valid else "NO VALID SCORES"
print("\n=== STUDENT SUMMARY ===", student_avg)

# Order students by average, highest first (tuple sort, no lambda)
ranking = [(avg, student) for student, avg in student_avg.items() if avg != "NO VALID SCORES"]
ranking.sort()
ranking.reverse()
print("=== ORDERED RESULTS ===", ranking)

print(f"\n=== FINAL REPORT ===")
print(f"Total: {len(records)} | Repeated: {len(repeated)} | Incomplete: {len(incomplete)}")
print(f"Top performer: {ranking[0]}")
```
