```python
from collections import defaultdict

LAST_NAME = "MEJOS"
SEED_NUM = 5
FAVORITE_ARTIST = "TAYLOR SWIFT"
PRODUCTS = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
PRICES = {"NOTEBOOK": 50, "PEN": 15, "USB": 250, "HEADPHONES": 800, "CALCULATOR": 120}

# Generate quantities using a list comprehension
quantities = [(ord(FAVORITE_ARTIST[i]) * SEED_NUM) % 10 + 1 for i in range(5)]
customers = [f"{LAST_NAME}_C{i + 1}" for i in range(5)]

# Build transactions as a list of tuples: (customer, product, quantity, value)
transactions = []
for i in range(5):
    qty = quantities[i]
    product = PRODUCTS[i]
    transactions.append((customers[i], product, qty, qty * PRICES[product]))

transactions.append((customers[0], PRODUCTS[0], 3, 3 * PRICES[PRODUCTS[0]]))  # repeated
transactions.append((customers[1], PRODUCTS[2], 2, 2 * PRICES[PRODUCTS[2]]))  # extra

print("=== GENERATED TRANSACTIONS ===")
for t in transactions:
    print(t)

by_customer = defaultdict(list)
for cust, prod, q, val in records:
    by_customer[cust].append((prod, q, val))

print("\n=== ORGANIZED TRANSACTION DATA ===")
for cust, entries in by_customer.items():
    print(cust, "->", entries)

# Repeated customer-product pairs using a set
seen = set()
repeated = []
for customer, product, qty, value in transactions:
    key = (customer, product)
    if key in seen:
        repeated.append((customer, product, qty, value))
    else:
        seen.add(key)

print("\n=== REPEATED ===", repeated)
print("=== DISTINCT PRODUCTS ===", set(t[1] for t in transactions))
print("=== DISTINCT CUSTOMERS ===", set(t[0] for t in transactions))

transaction_values = [val for cust, prod, q, val in records]
print("Transaction Values:", transaction_values)

# Totals per customer and product using defaultdict
customer_total = defaultdict(int)
product_total = defaultdict(int)
for customer, product, qty, value in transactions:
    customer_total[customer] += value
    product_total[product] += value

print("\n=== CUSTOMER SUMMARY ===", dict(customer_total))
print("=== PRODUCT SUMMARY ===", dict(product_total))

# Order transactions by value, highest first (sort by tuple, no lambda)
by_value = [(value, customer, product) for customer, product, qty, value in transactions]
by_value.sort()
by_value.reverse()
print("\n=== ORDERED RESULTS ===")
for value, customer, product in by_value:
    print(f"{customer} - {product}: {value}")

total = 0
for cust, prod, q, val in records:
    total += val
  print(f"\n=== OVERALL TRANSACTION SUMMARY ===")
  print(f"Total Transactions: {len(records)} | Repeated: {len(repeated)} | Total Value: {total}")
```
