```python
from collections import defaultdict

LAST_NAME = "MEJOS"
SEED_NUM = 5
FAVORITE_ARTIST = "TAYLOR SWIFT"
CATEGORIES = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

# Generate test sources: T1-T5 labeled with the surname
sources = [f"{LAST_NAME}_T{i + 1}" for i in range(5)]

# Generate records as a list of tuples: (source, category, value)
records = []
for i in range(5):
    value = (ord(FAVORITE_ARTIST[i]) * SEED_NUM) % 100
    records.append((sources[i], CATEGORIES[i], value))

# Additional measurements under at least two categories (same source, different category)
records.append((sources[0], "VOLTAGE", (ord(FAVORITE_ARTIST[5]) * SEED_NUM) % 100))
records.append((sources[1], "CURRENT", (ord(FAVORITE_ARTIST[6]) * SEED_NUM) % 100))

# Repeated test-source/category record
records.append((sources[0], "TEMPERATURE", records[0][2]))

print("=== GENERATED TEST DATA ===")
for r in records:
    print(r)

# Organize by source and by category using defaultdict
by_source = defaultdict(list)
by_category = defaultdict(list)
for source, category, value in records:
    by_source[source].append((category, value))
    by_category[category].append((source, value))

print("\n=== ORGANIZED BY SOURCE ===")
for source, entries in by_source.items():
    print(source, "->", entries)

print("=== ORGANIZED BY CATEGORY ===")
for category, entries in by_category.items():
    print(category, "->", entries)

# Identify repeated and distinct source/category pairs using a set
seen = set()
repeated = []
for source, category, value in records:
    key = (source, category)
    if key in seen:
        repeated.append((source, category, value))
    else:
        seen.add(key)

print("\n=== REPEATED DATA ===", repeated)
print("=== DISTINCT SOURCES ===", set(r[0] for r in records))
print("=== DISTINCT CATEGORIES ===", set(r[1] for r in records))

# Numerical and categorical summaries
all_values = [v for s, c, v in records]
avg_value = round(sum(all_values) / len(all_values), 2)
max_value = max(all_values)
min_value = min(all_values)

category_avg = {}
for category, entries in by_category.items():
    vals = [v for s, v in entries]
    category_avg[category] = round(sum(vals) / len(vals), 2)

print("\n=== NUMERICAL SUMMARY ===")
print(f"Average: {avg_value} | Max: {max_value} | Min: {min_value}")
print("=== CATEGORICAL SUMMARY (Avg per category) ===", category_avg)

# Order records by value, highest first (tuple sort, no lambda)
by_value = [(value, source, category) for source, category, value in records]
by_value.sort()
by_value.reverse()

print("\n=== ORDERED RESULTS (Highest Value First) ===")
for value, source, category in by_value:
    print(f"{source} - {category}: {value}")

print(f"\n=== FINAL ORGANIZED AND SUMMARIZED TEST DATASET ===")
print(f"Total Records: {len(records)} | Repeated: {len(repeated)} | Sources: {len(sources)} | Categories: {len(CATEGORIES)}")
print(f"Overall Average: {avg_value}")
```
